```python
import subprocess
import datetime

def auto_commit(message=None):
    # 1. Define a default message with a timestamp if none is provided
    if not message:
        timestamp = datetime.datetime.now().strftime("%Y-%m-%d %H:%M:%S")
        message = f"Auto-commit: Successful notebook run at {timestamp}"
    
    try:
        # 2. Stage all notebook changes
        subprocess.run(["git", "add", "."], check=True)
        
        # 3. Commit the changes
        subprocess.run(["git", "commit", "-m", message], check=True)
        
        # 4. Push to your main/master branch
        subprocess.run(["git", "push"], check=True)
        print(f"Pushed to Git: '{message}'")
        
    except subprocess.CalledProcessError as e:
        print(f"Git operation failed. Error details: {e}")

# Call the function at the end of your workflow
auto_commit()
```
